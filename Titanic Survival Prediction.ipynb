```python
cat_cols = df.select_dtypes(include = 'object').columns

cat_cols
```

```python
#label encoding 
#!pip install scikit-Learn
from sklearn.preprocessing import LabelEncoder
```

```python
#create an instance of our instances
le = LabelEncoder()
for col in cat_cols :
	df[col]=le.fit_transform(df[col])
    
df.head()
```

```python
#selecting useful feature
#remove unnecessary column
df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
df
```

```python
#model building 
df.head(2)
```

```python
#features
x = df.drop(columns =['Survived'])
#target
y= df['Survived']

#train 
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)

```

```python
pred_lr = lr.predict(x_test)

pred_lr
```

```python
from sklearn.metrics import r2_score

r2_score(y_test,pred_lr)
```

```python
#build all models using for loop 
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier



models={"Logistic Regression" :LogisticRegression(max_iter=1000),
	"DecisionTree":DecisionTreeClassifier(random_state=42),
	"RandomForest(100)":RandomForestClassifier(n_estimators=100,random_state=42),
	"RandomForest(50)":RandomForestClassifier(n_estimators=200,random_state=50)
}
```

```python
models

```

```python

models.values()



```

```python
models.items()
```

```python
from sklearn.metrics import accuracy_score
result={}

best_accuracy=0
best_model=None
best_model_name=""

for name,model in models.items():
	model.fit(x_train,y_train)
prediction=model.predict(x_test)
accuracy=accuracy_score(y_test,prediction)

result[name]=accuracy

print(f"{name} Accuracy :{accuracy}")
print(f"{name} Accuracy :{accuracy:2f}")
print(accuracy)

if accuracy > best_accuracy:
	best_accuracy=accuracy
	best_model=model
	best_model_name=name

print("\nBest MOdel:",best_model_name)
print("Best Accuracy:",best_accuracy)
```
